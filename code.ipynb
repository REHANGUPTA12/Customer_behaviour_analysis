```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("customer_shopping_behavior.csv")
df.head()
```

```python
df.columns = df.columns.str.lower().str.replace({
    " " : "_",
    "(" : "",
    ")" : ""
})
df.columns
```

```python
df.isna().sum()
```

```python
df.info()
```

```python
df["review_rating"] = df["review_rating"].fillna(
    df.groupby("category")["review_rating"].transform("median")
)
```

```python
df.isna().sum()
```

```python
df.info()
```

```python
labels = ["Young Adult", "Adult", "Middle-Aged", "Senior"]
df['age_group'] = pd.qcut(df['age'], q=4, labels= labels)
df['age_group']
```

```python
df['frequency_of_purchases'].unique().sum()
```

```python
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}
df['frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[ ['frequency_days', 'frequency_of_purchases']]
```

```python
df.to_csv("customer_data2.csv", index=False)
```
